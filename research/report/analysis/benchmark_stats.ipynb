```python
import json
import pandas as pd
from pathlib import Path
ROOT_DIR = Path(r"C:\repo\table-diversification")
```

```python
with open(r"C:\repo\table-diversification\research\dataset\overall_distorted_dataset\original.json", "r") as file:
    data = json.load(file)

rows = []
for d in data:
    df = pd.read_excel(ROOT_DIR / d['data_file'])
    rows.append(len(df))

print(f"Average number of rows: {sum(rows)/len(rows)}")
```
